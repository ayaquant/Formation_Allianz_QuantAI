```python
# NumPy

import numpy as np

# ------------------------------------------------------------
# 1. STATISTIQUES DE BASE
# ------------------------------------------------------------
Moyenne = np.mean([10, 20, 30])            # Moyenne
Médiane = np.median([10, 20, 30])          # Médiane
Ecart_type = np.std([10, 20, 30], ddof=1)     # Écart-type (échantillon)
Variance = np.var([10, 20, 30], ddof=1)     # Variance
Quantile = np.percentile([10, 20, 30], 99.5)  # Quantile (ex. VaR 99,5 % Solvabilité II)
Matrice_de_corrélation = np.corrcoef([1, 2, 3], [2, 4, 7])  # Matrice de corrélation

print(Moyenne)
print(Médiane)
print(Ecart_type)
print(Variance)
print(Quantile)
print(Matrice_de_corrélation)
```

```python
#SciPy

from scipy import stats, optimize, integrate
Probabilité = stats.norm.cdf(1.96)                    # Probabilité (≈ 0.975)
Quantile = stats.norm.ppf(0.995)                   # Quantile 99,5 % (≈ 2.576)
P_2décès = stats.binom.pmf(2, n=1000, p=0.002)     # P(2 décès sur 1000 assurés)
P_5_décès = stats.binom.sf(5, n=1000, p=0.002)      # P(plus de 5 décès)
P_120_décès= stats.poisson.sf(119, mu=100)           # P(≥ 120 décès | 100 attendus)
print(Probabilité)
print(Quantile)
print(P_2décès)
print(P_5_décès)
print(P_120_décès)
```
