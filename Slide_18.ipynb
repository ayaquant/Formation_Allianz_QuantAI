```python
#Données de l'exercice
import pandas as pd
import matplotlib.pyplot as plt

# 1. Données du portefeuille d'assurance décès (2025)
df = pd.DataFrame({
    "tranche": ["20-40", "40-55", "55-70", "70 et plus"],
    "taux": [1.2, 3.5, 9.0, 30.0]
})
```

```python
#Solution :
# Mise en forme :  Couleurs : tranche, taux (max en rouge)
couleurs = [
    "red" if t == df["taux"].max() else "steelblue"
    for t in df["taux"]
]
#Graphique
ax = df.plot.bar(
    x="tranche", y="taux",
    color=couleurs, rot=0
)

ax.set_ylabel("Taux de mortalité annuel (‰)")
ax.set_xlabel("Tranche d'âge (années)")

ax.axhline(5.0, linestyle="--")

ax.text(-0.4, 5.6, "moyenne 5,0 ‰")

plt.figtext(
    0.01, -0.05,
    "Période : exercice 2025 | Source : données fictives",
    fontsize=8, color="gray"
)

plt.show()
```
