```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Création du DataFrame
df = pd.DataFrame({
    "tranche_age": ["< 30 ans", "30-40", "40-50", "50-60", "60-70", "> 70 ans"],
    "taux_rachat": [6.8, 4.9, 3.6, 3.1, 4.4, 7.2]
})

# 2. Création du graphique
ax = df.plot.bar(
    x="tranche_age",
    y="taux_rachat",
    legend=False,
    figsize=(12, 6),
    color=["#AFC3DF", "#AFC3DF", "#AFC3DF", "#E8756B", "#AFC3DF", "#BF4934"]
)

# 3. Titres et axes
ax.set_title("Taux de rachat simulé par tranche d'âge — portefeuille vie synthétique",
             fontsize=15, fontweight="bold")
ax.set_xlabel("Tranche d'âge à la date d'observation (années)", fontsize=12)
ax.set_ylabel("Taux de rachat annuel (%)", fontsize=12)

# 4. Ligne de moyenne
moyenne = df["taux_rachat"].mean()
ax.axhline(moyenne, color="navy", linestyle="--", linewidth=2,
           label=f"Moyenne {moyenne:.1f} %")
ax.legend()

# 5. Affichage des valeurs au-dessus des barres
for i, valeur in enumerate(df["taux_rachat"]):
    ax.text(i, valeur + 0.15, f"{valeur:.1f} %",
            ha="center", fontweight="bold", fontsize=11, color="navy")

# 6. Grille et mise en page
ax.set_ylim(0, 8.6)
ax.grid(axis="y", alpha=0.3)
ax.set_axisbelow(True)

# 7. Période et source
plt.figtext(
    0.02, -0.04,
    "Période : rachats totaux observés sur l'exercice 2025 (taux annuel, en % des contrats en début d'année)\n"
    "Source : simulation interne, portefeuille épargne vie synthétique (données fictives, support de formation)",
    fontsize=9, color="gray"
)

plt.tight_layout()
plt.show()

```
